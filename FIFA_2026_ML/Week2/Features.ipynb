```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("features.csv")

# Initialize each team's starting Elo using the existing ratings
current_elo = {}
teams = pd.unique(df[['team1', 'team2']].values.ravel())

for t in teams:
    first_team1 = df.loc[df['team1'] == t, 'team1_elo']
    first_team2 = df.loc[df['team2'] == t, 'team2_elo']
    if not first_team1.empty:
        current_elo[t] = float(first_team1.iloc[0])
    elif not first_team2.empty:
        current_elo[t] = float(first_team2.iloc[0])
    else:
        current_elo[t] = 1500.0  # fallback (very rare)

# Elo constants
K = 40.0  # sensitivity factor

records = []

# Compute match-by-match features
for _, row in df.iterrows():
    t1, t2 = row['team1'], row['team2']
    s1, s2 = row['team1_score'], row['team2_score']

    # Current Elo ratings
    E1 = current_elo.get(t1, 1500.0)
    E2 = current_elo.get(t2, 1500.0)

    # Expected probability (no home advantage)
    exp1 = 1 / (1 + 10 ** ((E2 - E1) / 400))
    exp2 = 1 - exp1

    # Actual results
    if s1 > s2:
        S1, S2 = 1, 0
    elif s1 < s2:
        S1, S2 = 0, 1
    else:
        S1, S2 = 0.5, 0.5

    # Goal margin multiplier
    goal_diff = abs(int(s1) - int(s2))
    G = np.log(goal_diff + 1) * 0.5

    # Elo updates
    E1_new = E1 + K * G * (S1 - exp1)
    E2_new = E2 + K * G * (S2 - exp2)

    # Store results
    records.append({
        "team1": t1,
        "team2": t2,
        "team1_elo_before": E1,
        "team2_elo_before": E2,
        "elo_diff_before": E1 - E2,
        "expected_team1_win": exp1,
        "goal_margin_factor": G,
        "team1_elo_after": E1_new,
        "team2_elo_after": E2_new,
        "elo_change_team1": E1_new - E1,
        "elo_change_team2": E2_new - E2,
        "result_numeric": row.get("result_numeric", np.nan)
    })

    # Update Elo ratings for next iteration
    current_elo[t1] = E1_new
    current_elo[t2] = E2_new

# Build the new DataFrame
elo_features = pd.DataFrame(records)

# Display first few rows
print(elo_features.head())

# Save to CSV
elo_features.to_csv("elo_features_O.csv", index=False)

```
